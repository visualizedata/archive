## Data Key Definitions

Fill out the right hand side of the associations below.

For `COLLECTION_DEFINITIONS` the script expects a course name (one of: "thesis", "ms1", "ms2" or "other") and a 4-digit year.

For `DATA_DEFINITIONS` the right hand side should be the name of the columns in the spreadsheet.

The `project_tags` definition is the only one that expects a list of column names.

```python
COLLECTION_DEFINITIONS = {
  "course": "",
  "year": None,
}

DATA_DEFINITIONS = {
  "first_name": "",
  "last_name": "",
  "project_title": "",
  "project_subtitle": "",
  "project_description": "",
  "project_tags": ["", "", ""],
  "repo_url": "",
  "project_url": "",
  "student_url": "",
  "image_url": "",
}

check_collection_definitions(COLLECTION_DEFINITIONS)
check_data_definitions(sheet_df, DATA_DEFINITIONS, optional=["project_subtitle"])

extracted_df = extract_data(sheet_df, DATA_DEFINITIONS, COLLECTION_DEFINITIONS, optional=["project_subtitle"])
extracted_df
```
